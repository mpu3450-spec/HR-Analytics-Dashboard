# Outliers Detection

```python
q3 = df['Age'].quantile(0.75)
q1 = df['Age'].quantile(0.25)
iqr = q3 - q1
```

```python
high = q3 + 1.5 * iqr
low = q1 - 1.5 * iqr
```

```python
# Outliers
df[(df['Age'] > high) | (df['Age'] < low)]
#NO OUTLIER PRESENT
```
